# Messages

Messages are the fundamental unit of context for models in langchain. They represent the input and output of models, carrying both the content and metadata to represent the state of a consersation when interacting with an LLM. Messages are objects that contain: 


- Role : Identifies the message type (ef: System, user)
- Cpntent - Represents the actual content of the message (like text, images, audio, documents, etc.)
- Metadata : Optional fields such as response information, message ID's and token usage

Langchain provides a standard message type that works across all model providers, ensuring consistent behaviour regardless of the model being called. 

```python
import os 
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-20b")
```

```python
model.invoke("Please tell me if coding is hard or easy ?")
```

# Text Prompts

Text prompts are strings - ideal for straightforward generation tasks where you don't need to retain converstation history.

```python
model.invoke("you are the best")
```

# Message prompt

Alternative, you can pass in list of messages to the model by providing a list of message objects. 

Message type:

- system message - Tells the model how to behave and provide contect for interactions
- Human message - Represents user input and interactions with model 
- AI message - Respnses generated by the model, including text context,  tool call, and metadata
- Tool Message - Repreesents the outputs of tool calls

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

message = [ 
    SystemMessage("you are a poetry expert"),
    HumanMessage("write a poem onartificial intelligence")
]

response = model.invoke(message)
response.content
```

```python
system_msg = SystemMessage("You are a helpful coding assistance")

messages = [
    system_msg, 
    HumanMessage("How do I create a Rest API?")
]

response = model.invoke(messages)
print(response.content)
```

```python
## Detailed info to the LLM theough System message

from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("""
You are a senior Python developer with expertise in web frameworks.
Always provide code examples and explain your reasoing.
Be concise but thorough in your explanations.
""")


message = [
    system_msg, 
    HumanMessage("How do i create a rest api")
]

response = model.invoke(message)
print(response.content)
```

```python
## Messgae Metadata

human_msg = HumanMessage(
    content = "Hello",
    name = "alice",
    id="msg_123"
)
```

```python
response = model.invoke([
    human_msg
])
response
```

```python
from langchain.messages import AIMessage, SystemMessage, HumanMessage

# Create an AI message manually (e.g., for conversation history)
ai_msg = AIMessage("I'd be happy to help you with that question!")

# Add to conversation history
messages = [
    SystemMessage("You are a helpful assistant and explains why the result is what it is !"),
    HumanMessage("Can you help me?"),
    ai_msg,  # Insert as if it came from the model
    HumanMessage("Great! What's 2+2?")
]

response = model.invoke(messages)
print(response.content)
```

```python
response.usage_metadata
```

```python
from langchain.messages import AIMessage, ToolMessage

ai_message = AIMessage(
    content= [],
    tool_calls=[{
        "name": "get_weather",
        "args" : {"location":"San Francisco"},
        "id": "calls_123"
    }]
)



#Execute tool and create result message
weather_result = "sunny, 72F"
tool_message = ToolMessage(
    content= weather_result,
    tool_call_id= "call_123"
)

#Continue conversation

message = [
    HumanMessage("what's the weather in San Feancisco?"),
    ai_message, 
    tool_message
]

response = model.invoke(message)
```
